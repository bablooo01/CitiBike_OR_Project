```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
DATA_DIR = Path(r"D:\CitiBike_OR_Project\data\raw")

files = sorted(DATA_DIR.glob("*.csv"))

print("Number of CSV files:", len(files))

for file in files:
    print(file.name)
```

```python
for file in files:
    df = pd.read_csv(file, nrows=5)

    print("\n" + "=" * 60)
    print(file.name)
    print("=" * 60)

    print("Columns:")
    print(df.columns.tolist())

    print("Shape of sample:", df.shape)
```

```python
def count_rows(file):
    count = 0

    for chunk in pd.read_csv(file, usecols=["ride_id"], chunksize=100_000):
        count += len(chunk)

    return count


for file in files:
    print(file.name, ":", f"{count_rows(file):,} rows")
```
